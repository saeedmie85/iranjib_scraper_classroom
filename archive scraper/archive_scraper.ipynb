```python
import requests
from bs4 import BeautifulSoup
i = 0
urls = []
while True:
    i += 1
    archive_url = f'https://www.iranjib.ir/jax/showarchive.php?p=1&_id={i}'
    page_response = requests.get(archive_url)
    page_soup = BeautifulSoup(page_response.text, 'html.parser')
    tabel_soup = page_soup.findAll('table')[0]
    # tbody_soup = tabel_soup.find('tbody')
    trs_soup = tabel_soup.find_all('tr')
    for tr_soup in trs_soup:
        title = tr_soup.find_all('td')[0].find('a').get_text().strip()
        link = tr_soup.find_all('td')[0].find('a').get('href')
        date = tr_soup.find_all('td')[-1].get_text()
        if 'آبان' in date:
            break
        if ('طلا' in title) or ('دلار' in title):
            urls.append(link)

    if 'آبان' in date:
            break
```

```python
for url in urls:
    scrap_page(url)
```
