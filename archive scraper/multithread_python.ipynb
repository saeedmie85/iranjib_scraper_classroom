```python
import threading
import time

def square(num):
    for _ in range(10):
        print(f"Square: {num*num}")
        time.sleep(1)

def cube(num):
    for _ in range(10):
        print(f"Cube: {num*num*num}")
        time.sleep(3)

# square(10)
# cube(5)
t1 = threading.Thread(target=square, args=(10,))
t2 = threading.Thread(target=cube, args=(5,))
t3 = threading.Thread(target=square, args=(7,))

t1.start()
t2.start()
t3.start()
# t1.join()
# t2.join()

print("Done!")
```
